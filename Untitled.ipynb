```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
from datetime import datetime
from fake_useragent import UserAgent

print('All imports successful!')
```

```python
ua = UserAgent()

def get_headers():
    return {
        'User-Agent': ua.random,
        'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
        'Accept-Language': 'en-IN,en;q=0.9',
        'Accept-Encoding': 'gzip, deflate, br',
        'Connection': 'keep-alive',
        'Upgrade-Insecure-Requests': '1',
        'Referer': 'https://www.amazon.in/',
    }

print('Headers setup done!')
```

```python
response = requests.get(
    'https://www.amazon.in/s?k=laptops&page=1',
    headers=get_headers(),
    timeout=15
)

soup = BeautifulSoup(response.content, 'html.parser')
cards = soup.find_all('div', {'data-component-type': 's-search-result'})
print(f'Status code: {response.status_code}')
print(f'Total cards: {len(cards)}')

if cards:
    card = cards[0]

    # Title
    h2 = card.find('h2')
    print(f'\nTitle (h2): {h2.get_text(strip=True)[:80] if h2 else "NOT FOUND"}')

    # Price
    price = card.find('span', {'class': 'a-price-whole'})
    print(f'Price: {price.get_text(strip=True) if price else "NOT FOUND"}')

    # Rating
    rating = card.find('span', {'class': 'a-icon-alt'})
    print(f'Rating: {rating.get_text(strip=True) if rating else "NOT FOUND"}')

    # Image
    img = card.find('img', {'class': 's-image'})
    print(f'Image: {img["src"][:60] if img else "NOT FOUND"}')

    # Ad or Organic
    classes = card.get('class', [])
    print(f'Ad/Organic: {"Ad" if "AdHolder" in classes else "Organic"}')
```

```python
def scrape_amazon_laptops(pages=3):
    base_url = 'https://www.amazon.in/s?k=laptops&page={}'
    all_products = []

    for page in range(1, pages + 1):
        print(f'\nScraping page {page}...')
        url = base_url.format(page)

        try:
            response = requests.get(url, headers=get_headers(), timeout=15)
            print(f'  Status code: {response.status_code}')

            if response.status_code != 200:
                print(f'  Skipping page {page}')
                continue

            soup = BeautifulSoup(response.content, 'html.parser')
            cards = soup.find_all('div', {'data-component-type': 's-search-result'})
            print(f'  Found {len(cards)} product cards')

            for card in cards:
                product = {}

                # Ad or Organic
                classes = card.get('class', [])
                product['Result Type'] = 'Ad' if 'AdHolder' in classes else 'Organic'

                # Title
                h2 = card.find('h2')
                product['Title'] = h2.get_text(strip=True) if h2 else 'N/A'

                # Price
                price = card.find('span', {'class': 'a-price-whole'})
                product['Price (INR)'] = price.get_text(strip=True).replace(',', '') if price else 'N/A'

                # Rating
                rating = card.find('span', {'class': 'a-icon-alt'})
                product['Rating'] = rating.get_text(strip=True) if rating else 'N/A'

                # Image
                img = card.find('img', {'class': 's-image'})
                product['Image URL'] = img['src'] if img else 'N/A'

                if product['Title'] != 'N/A':
                    all_products.append(product)

        except requests.exceptions.RequestException as e:
            print(f'  Request error on page {page}: {e}')

        delay = random.uniform(2, 5)
        print(f'  Waiting {delay:.1f}s before next page...')
        time.sleep(delay)

    return all_products

print('Scraper function defined!')
```

```python
print('Starting scraper...')
print('=' * 50)

products = scrape_amazon_laptops(pages=3)

print('\n' + '=' * 50)
print(f'Total products scraped: {len(products)}')
```

```python
if products:
    df = pd.DataFrame(products)

    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    filename = f'amazon_laptops_{timestamp}.csv'

    df.to_csv(filename, index=False, encoding='utf-8-sig')

    print(f'Saved to: {filename}')
    print(f'Total rows: {len(df)}')
    display(df.head(5))
else:
    print('No products scraped.')
```

```python
if products:
    print('===== SCRAPE SUMMARY =====')
    print(f"Total Products  : {len(df)}")
    print(f"Ads             : {len(df[df['Result Type'] == 'Ad'])}")
    print(f"Organic         : {len(df[df['Result Type'] == 'Organic'])}")
    print(f"With Rating     : {len(df[df['Rating'] != 'N/A'])}")
    print(f"With Price      : {len(df[df['Price (INR)'] != 'N/A'])}")
    print(f"Saved File      : {filename}")
```
